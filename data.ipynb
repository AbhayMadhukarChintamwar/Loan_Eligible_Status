```python
import pandas as pd
import numpy as np

np.random.seed(42)

n = 1000

df = pd.DataFrame({
    "Loan_ID": [f"LP{100001+i}" for i in range(n)],

    "Gender": np.random.choice(
        ["Male", "Female"], n, p=[0.65, 0.35]
    ),

    "Married": np.random.choice(
        ["Yes", "No"], n, p=[0.65, 0.35]
    ),

    "Dependents": np.random.choice(
        ["0", "1", "2", "3+"], n,
        p=[0.58, 0.17, 0.17, 0.08]
    ),

    "Education": np.random.choice(
        ["Graduate", "Not Graduate"], n,
        p=[0.72, 0.28]
    ),

    "Self_Employed": np.random.choice(
        ["Yes", "No"], n,
        p=[0.15, 0.85]
    ),

    "ApplicationIncome": np.random.lognormal(
        mean=np.log(5000),
        sigma=0.55,
        size=n
    ).round().astype(int),

    "CoapplicantIncome": np.random.lognormal(
        mean=np.log(1500),
        sigma=0.75,
        size=n
    ).round().astype(int),

    "LoanAmount": np.random.lognormal(
        mean=np.log(140),
        sigma=0.40,
        size=n
    ).round().astype(int),

    "Loan_Amount_Term": np.random.choice(
        [120, 180, 240, 300, 360, 480],
        n,
        p=[0.03, 0.04, 0.06, 0.08, 0.72, 0.07]
    ),

    "Credit_History": np.random.choice(
        [1.0, 0.0], n,
        p=[0.82, 0.18]
    ),

    "Property_Area": np.random.choice(
        ["Urban", "Semi Urban", "Rural"], n,
        p=[0.32, 0.38, 0.30]
    )
})

# -------------------------------
# Target variable
# -------------------------------

score = (
    0.9 * df["Credit_History"]
    + 0.20 * (df["Education"] == "Graduate")
    + 0.12 * (df["Married"] == "Yes")
    + 0.10 * (df["Property_Area"] == "Semi Urban")
    + 0.08 * (
        df["ApplicationIncome"]
        > df["ApplicationIncome"].median()
    )
    - 0.20 * (
        df["LoanAmount"]
        > df["LoanAmount"].quantile(0.75)
    )
    + np.random.normal(0, 0.35, n)
)

df["Loan_Status"] = np.where(
    score > 0.55,
    "Approved",
    "Rejected"
)

# -------------------------------
# Add a small amount of NaN values
# -------------------------------

missing_rates = {
    "Gender": 0.012,
    "Married": 0.010,
    "Dependents": 0.015,
    "Self_Employed": 0.012,
    "CoapplicantIncome": 0.010,
    "LoanAmount": 0.012,
    "Loan_Amount_Term": 0.010,
    "Credit_History": 0.015,
    "Property_Area": 0.010
}

for column, rate in missing_rates.items():

    number_of_missing = int(n * rate)

    indexes = np.random.choice(
        n,
        number_of_missing,
        replace=False
    )

    df.loc[indexes, column] = np.nan

# -------------------------------
# Save CSV
# -------------------------------

df.to_csv("LoanData.csv", index=False)

print("LoanData.csv created successfully!")
print("Shape:", df.shape)

print("\nMissing values:")
print(df.isna().sum())

print("\nFirst 10 rows:")
print(df.head(10))
```
